```python
import requests
import pandas as pd
import sqlite3
import time
API_KEY = "130daf9c70a975ecdcd243776686cd94"

url = "https://api.themoviedb.org/3/discover/movie"
params = {
    "api_key": API_KEY,
    "language": "en-US",
    "page": 1
}
```

```python
movies = []

for page in range(1, 3):
    params["page"] = page
    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()
        movies.extend(data['results'])

    time.sleep(1)
```

```python
movies = []

for page in range(1, 3):
    params["page"] = page
    response = requests.get(url, params=params)

    print(response.status_code)

    if response.status_code == 200:
        data = response.json()
        movies.extend(data.get('results', []))
    else:
        print(response.text)

    time.sleep(1)
```

```python
movies_df = movies_df[['id', 'title', 'popularity', 'vote_average', 'release_date']]

movies_df.rename(columns={'id': 'movie_id'}, inplace=True)

movies_df.head()
```

```python
import sqlite3

conn = sqlite3.connect('movies.db')

movies_df.to_sql('movies', conn, if_exists='replace', index=False)

conn.close()
```
